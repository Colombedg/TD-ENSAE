```python
import pandas as pd

emissions = pd.read_csv("https://koumoul.com/s/data-fair/api/v1/datasets/igt-pouvoir-de-rechauffement-global/convert")
emissions.head(2)
```

```python
emissions_copy = emissions[['INSEE commune', 'Commune', 'Autres transports international', 'Autres transports']].copy()
emissions_copy.head()
```

```python
emissions_copy = emissions_copy.rename({"INSEE commune": "code_insee", "Autres transports": "transports", "Autres transports international": "transports_international"}, axis=1)
emissions_copy.head()
```

```python
emissions_copy.fillna(0, inplace=True)
emissions_copy["transports"]=emissions_copy["transports"].astype(int)
emissions_copy.head()
```
